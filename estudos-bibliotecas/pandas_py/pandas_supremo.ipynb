```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv("vendas_carros_cidades.csv")
# pd.read_excel("gerentes_lojas_faturamento.xlsx")

display(df_vendas)
```

```python
#Inspenção
display(df_vendas) #As 5 primeiras linhas e ultimas linhas (No caso todas as linhas)
#display(df_vendas.head(2)) #As 5 primeiras linhas
#display(df_vendas.tail(2)) #As 2 ultimas linhas
# display(df_vendas.sample(3)) #3 linhas aleatorias
# display(df_vendas.shape) #Quantidade de linhas e colunas
# display(df_vendas.columns) #Todas as colunas
# display(df_vendas.info()) #Informações sobre o DataFrame
display(df_vendas.describe()) #Estatisticas descritivas do DataFrame
```

```python
##Tratamento
df_analise = df_vendas.copy() ##Criando uma copia do DataFrame original


##Colunas
# display(df_analise[["Cidade", "Valor_Total_Vendas"]]) ##Selecionando colunas especificas
df_analise = df_vendas.drop(columns=["Valor_Total_Vendas"]) ##Dropando colunas especificas

##Nulos
# df_analise = df_vendas.fillna(0) ##Substituindo valores nulos por 0
df_analise = df_vendas.dropna() ##Dropando valores nulos

#Tipos de dados
# df_analise["Data"] = pd.to_datetime(df_analise["Data"], format='%Y-%m-%d') #Caso houvesse uma coluna de data esse comando iria formata-lá para o tipo datetime


##Padronização
df_analise["Cidade"] = df_analise["Cidade"].str.title() ##Padronizando para a primeira letra maiuscula
df_analise["Cidade"] = df_analise["Cidade"].str.strip() ##Removendo espaços em branco no inicio e no final da string

##Duplicatas
# df_analise = df_analise.drop_duplicates() ##Dropando duplicatas
# df_analise = df_analise.drop_duplicates(subset=["Cidade"]) ##Dropando duplicatas com base em uma coluna especifica, Se duas cidades forem iguais, apenas a primeira será mantida



display(df_analise)
display(df_analise.info()) #Informações sobre o DataFrame
```

```python
##Criando colunas

##Faturamento
df_analise["Preço_Unitario"] = df_analise["Valor_Total_Vendas"] / df_analise["Quantidade_Vendida"] ##Criando uma nova coluna com base em outras colunas


##Forma de Venda
df_analise["Forma_de_Venda"] = np.where(df_analise["Quantidade_Vendida"] > 1, "Venda em Lote", "Venda Individual") ##Criando uma nova coluna com base em uma condição

##Região
# display(df_analise["Cidade"].unique()) ##Exibindo valores unicos da coluna Cidade

dic_regioes ={
    'São Bernardo Do Campo - Sp (Sbc)': 'Sudeste',
    'Rio De Janeiro - Rj (Rj)': 'Sudeste',
    'Santa Catarina - Sc': 'Sudeste',
    'Teresina - Pi (The)': 'Nordeste'
}

df_analise["Região"] = df_analise["Cidade"].map(dic_regioes) ##Criando uma nova coluna com base em um dicionario

display(df_analise)
display(df_analise.info()) #Informações sobre o DataFrame
display(df_analise.isna().sum()) #Exibindo a quantidade de valores nulos por coluna
```

```python
##Analisar -> filtrar

##Ordenando a lista
df_analise = df_analise.sort_values(by="Quantidade_Vendida", ascending=False) ##Ordenando o DataFrame com base em uma coluna especifica
df_analise = df_analise.reset_index(drop=True) ##Resetando o index do DataFrame
##Modos: loc, iloc, condicional

##1: loc || Pesquisando por nome da linha e nome da coluna
cidade = df_analise.loc[2, "Cidade"] ##Selecionando uma linha e coluna especifica
carro = df_analise.loc[2, "Carro"] ##Selecionando uma linha e coluna especifica
preco_unitario = df_analise.loc[2, "Preço_Unitario"] ##Selecionando uma linha e coluna especifica
print(f"A cidade é: {cidade}, o carro é: {carro} e o preço unitario é: {preco_unitario}")

##2: iloc || posição da linha e posição da coluna
cidade = df_analise.iloc[2, 0] ##Selecionando uma linha e coluna especifica
carro = df_analise.iloc[2, 1] ##Selecionando uma linha e coluna especifica
preco_unitario = df_analise.iloc[2, 4] ##Selecionando uma linha e coluna especifica
print(f"A cidade é: {cidade}, o carro é: {carro} e o preço unitario é: {preco_unitario}")

##3: Condicional || Pesquisando por uma condição especifica
df_analise_sbc = df_analise[df_analise["Cidade"] == "São Bernardo Do Campo - Sp (Sbc)"] ##Selecionando todas as linhas onde a coluna Cidade é igual a "São Bernardo Do Campo - Sp (Sbc)"

##Podemos usar o loc com a condicional para selecionar uma linha e coluna especifica
cidade = df_analise.loc[df_analise["Cidade"] == "São Bernardo Do Campo - Sp (Sbc)", "Cidade"].values[0] ##Selecionando uma linha e coluna especifica com base em uma condição e pegando somente os valores da coluna Cidade
carro = df_analise.loc[df_analise["Cidade"] == "São Bernardo Do Campo - Sp (Sbc)", "Carro"].values[0] ##Selecionando uma linha e coluna especifica
preco_unitario = df_analise.loc[df_analise["Cidade"] == "São Bernardo Do Campo - Sp (Sbc)", "Preço_Unitario"].values[0] ##Selecionando uma linha e coluna especifica
print(f"A cidade é: {cidade}, o carro é: {carro} e o preço unitario é: {preco_unitario}")


##4: Duplas condições
condicao1 = df_analise["Valor_Total_Vendas"]>2000000
condicao2 = df_analise["Região"]=="Sudeste"
df_analise_vendas_2mi = df_analise[condicao1 & condicao2]


display(df_analise)
display(df_analise_sbc)
df_analise_sbc.to_csv("Vendas_SBC.csv", index=False) ##Exportando o DataFrame para um arquivo CSV
display(df_analise_vendas_2mi)
```

```python
##Analise -> Agrupamento | grupby()
display(df_analise)

#Ranking de faturamento por loja
analise_lojas = df_analise[["Cidade", "Valor_Total_Vendas"]].groupby("Cidade").sum()
analise_lojas = analise_lojas.sort_values("Valor_Total_Vendas", ascending=False)
analise_lojas = analise_lojas.reset_index()
analise_lojas["Valor_Total_Vendas"] = analise_lojas["Valor_Total_Vendas"].map("R${:,.2f}".format)
analise_lojas = analise_lojas.rename(columns={"Valor_Total_Vendas": "Valor Total"})
display(analise_lojas)

analise_qnt_lojas = df_analise[["Cidade", "Quantidade_Vendida"]].groupby("Cidade").sum()
analise_qnt_lojas = analise_qnt_lojas.sort_values("Quantidade_Vendida", ascending=False)
analise_qnt_lojas = analise_qnt_lojas.reset_index()
analise_qnt_lojas = analise_qnt_lojas.rename(columns={"Quantidade_Vendida": "Quantidade de Vendas"})
display(analise_qnt_lojas)

```
